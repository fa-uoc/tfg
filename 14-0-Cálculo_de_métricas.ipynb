# Cálculo de score para selección de aumentaciones

Cargamos csv con las métricas de los experimentos anteriores y de la besline:

```python
import numpy as np
import pandas as pd
df = pd.read_csv("metrics/metrics.csv")
df.columns = df.columns.str.strip()

baseline = df.iloc[0]


df = df.copy()
df["delta_f1"] = df["F1"] - baseline["F1"]
df["delta_etime"] = df["Epoch time"] - baseline["Epoch time"]

def normalize(s):
    return (s - s.min()) / (s.max() - s.min())

df["norm_gain"] = normalize(df["delta_f1"])
df["norm_cost"] = normalize(df["delta_etime"])

w_gain = 0.9
w_cost = 0.1
df["puntuación"] = w_gain * df["norm_gain"] - w_cost * df["norm_cost"]

result = df.sort_values("puntuación", ascending=False)
result
```
